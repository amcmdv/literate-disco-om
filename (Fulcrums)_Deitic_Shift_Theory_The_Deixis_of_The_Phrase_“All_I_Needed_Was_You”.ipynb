# Deixis : Analysising Wordcraft

The phrase “all I needed was you” is deictically rich: it indexes the **speaker (I)**, the **addressee (you)**, and a **past time (needed)**. It relies on the shared context to specify who “you” is, and the social deixis frames the relationship as **direct, intimate, and exclusive**.

We shall slowly work towards the relation with 'Jamdani'

### 1. Person deixis

* **“I”** → **first person singular**; the speaker is indexing themselves.
* **“you”** → **second person singular/plural**; the addressee is directly invoked.
  → This creates an **interpersonal alignment**: the utterance positions the relationship between speaker and addressee as the central frame.

### 2. Time deixis

* **“needed”** is past tense, anchoring the utterance to a time **before the moment of speaking**.
* The deictic center is the **present moment of narration**, with the need located in the past.

### 3. Discourse deixis

* The phrase implies contrast with other possible things/people. The **“all”** restricts the set: in the discourse, this excludes other resources, options, or people.

### 4. Social deixis

* By using “you” (not a third-person reference), the speaker **foregrounds intimacy or directness**. The **absence of modifiers** (“sir,” “doctor,” “my friend”) suggests solidarity or closeness.
* Depending on context (song lyric, conversation, letter), “you” could encode a romantic partner, a friend, a deity, or even an audience.


```
import networkx as nx
import plotly.graph_objects as go
import numpy as np

# ==========================================
# 1. FORENSIC PARAMETERS & PHYSICS SETUP
# ==========================================
# These parameters represent the linguistic forces at play.
# Adjusting these will change the physical structure of the sentence.

SOCIAL_DEIXIS_TENSION = 5.0    # High tension = Intimate, short distance between "I" and "You"
DISCOURSE_REPULSION = 2.5      # Strength of "All" pushing away external options
TIME_DEIXIS_GRAVITY = -1.0     # Pulls "Needed" to the left (negative X axis representing the past)

# Initialize the graph
G = nx.Graph()

# Define nodes and their semantic categories
core_nodes = ['I', 'You', 'Needed', 'All']
excluded_nodes = ['Other Person A', 'Other Option B', 'External Context']
G.add_nodes_from(core_nodes + excluded_nodes)

# Define the physical springs (edges)
# Weight determines the resting length and stiffness of the spring
G.add_edge('I', 'You', weight=SOCIAL_DEIXIS_TENSION)
G.add_edge('I', 'Needed', weight=2.0)
G.add_edge('You', 'Needed', weight=2.0)
G.add_edge('Needed', 'All', weight=3.0)

# ==========================================
# 2. COMPUTE FORCE-DIRECTED LAYOUT
# ==========================================
# Uses the Fruchterman-Reingold algorithm to simulate anti-gravity (repulsion)
# between all nodes and spring attraction along edges.
pos = nx.spring_layout(G, k=1/DISCOURSE_REPULSION, iterations=100, weight='weight', seed=42)

# Apply Time Deixis (Directional Force)
# Forcefully anchor "Needed" to the past (left side of the X-axis)
pos['Needed'][0] += TIME_DEIXIS_GRAVITY

# Apply Discourse Deixis (Repulsion)
# Forcefully scatter the excluded nodes away from the central deictic cluster
for i, node in enumerate(excluded_nodes):
    pos[node] = np.array([1.5 + np.random.rand(), (i - 1) * 1.5 + np.random.rand()])

# ==========================================
# 3. INTERACTIVE PLOTLY VISUALIZATION
# ==========================================
edge_x = []
edge_y = []
for edge in G.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

edge_trace = go.Scatter(
    x=edge_x, y=edge_y,
    line=dict(width=2, color='#888'),
    hoverinfo='none',
    mode='lines')

node_x = []
node_y = []
node_text = []
node_color = []

for node in G.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)
    node_text.append(node)

    # Color coding based on semantic role
    if node in ['I', 'You']:
        node_color.append('#1f77b4') # Blue for Person Deixis
    elif node == 'Needed':
        node_color.append('#ff7f0e') # Orange for Time Deixis
    elif node == 'All':
        node_color.append('#2ca02c') # Green for Discourse Deixis
    else:
        node_color.append('#000000') # Red for Excluded Elements

node_trace = go.Scatter(
    x=node_x, y=node_y,
    mode='markers+text',
    text=node_text,
    textposition="top center",
    hoverinfo='text',
    marker=dict(
        showscale=False,
        color=node_color,
        size=35,
        line_width=2))

# Render the interactive figure
fig = go.Figure(data=[edge_trace, node_trace],
             layout=go.Layout(
                title='Deictic Force Simulation: "All I needed was you"',
                titlefont_size=16,
                showlegend=False,
                hovermode='closest',
                margin=dict(b=20,l=5,r=5,t=40),
                xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
                yaxis=dict(showgrid=False, zeroline=False, showticklabels=False))
                )

fig.show()
```

### Forensic Breakdown of the Architecture

* **`SOCIAL_DEIXIS_TENSION` (Spring Constant):** In `NetworkX`, passing this variable as the `weight` parameter into the edge connecting 'I' and 'You' dictates the stiffness of the simulated spring. Higher tension (intimacy) restricts the layout algorithm from drifting these nodes apart.
* **`DISCOURSE_REPULSION` (Anti-Gravity):** The `k` parameter in `nx.spring_layout` dictates optimal distance between nodes. By inversely tying this to our discourse variable, we simulate the restrictive nature of "All," pushing unlinked nodes (the excluded options) to the periphery.
* **`TIME_DEIXIS_GRAVITY` (Vector Shift):** Post-simulation, we manually apply a vector transformation (`pos['Needed'][0] += TIME_DEIXIS_GRAVITY`) to physically drag the temporal anchor backward along the x-axis, visualizing the past-tense anchoring against the present-tense cluster.
